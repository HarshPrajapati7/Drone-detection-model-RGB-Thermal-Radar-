### 2. Visualize Bounding Boxes (for PPT Export)

Once the inference is complete, run the cell below to load the output images with the bounding boxes overlayed so you can screenshot them directly for the PPT.

```python
import os
import glob
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Path to the inferred images output by TAO
results_dir = r"F:\Sih26change_Data\runs\inference_test\images"

if os.path.exists(results_dir):
    image_files = glob.glob(os.path.join(results_dir, "*.jpg")) + glob.glob(os.path.join(results_dir, "*.png"))
    
    if not image_files:
        print("No images found in the output directory. Did inference succeed?")
    else:
        # Display the first 4 output images
        plt.figure(figsize=(15, 10))
        for i, img_path in enumerate(image_files[:4]):
            img = mpimg.imread(img_path)
            plt.subplot(2, 2, i+1)
            plt.imshow(img)
            plt.axis('off')
            plt.title(f"Inference Result {i+1}")
        plt.tight_layout()
        plt.show()
else:
    print(f"Directory does not exist yet: {results_dir}")
```
